Q1. Write a unique paragraph (5-6 sentences) about your favorite topic (e.g., sports,
technology, food, books, etc.).
1. Convert text to lowercase and remove punctua5on using re.
2. Tokenize the text into words and sentences.
3. Split using split() and word_tokenize() and compare how Python split and NLTK’s
word_tokenize() differ.
4. Remove stopwords (using NLTK's stopwords list).
5. Display word frequency distribu5on (excluding stopwords).

```python

# STEP 1: Import Libraries
import re
import nltk
import matplotlib.pyplot as plt

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# Download required NLTK resources
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

print("=" * 60)
print("Q1: TEXT PREPROCESSING AND WORD FREQUENCY")
print("=" * 60)


# STEP 2: Write a Unique Paragraph (6 Sentences)
text = """
Artificial intelligence is changing the way people work and learn.
It helps computers understand information and solve complex problems.
AI is used in healthcare, education, transportation, and many other fields.
Machine learning allows systems to improve their performance using data.
AI tools can save time and support people in making better decisions.
I believe artificial intelligence will play an important role in the future.
"""

print("\n1. ORIGINAL PARAGRAPH:")
print(text)


# STEP 3: Convert to Lowercase and Remove Punctuation
lower_text = text.lower()

# Remove punctuation using regular expressions
clean_text = re.sub(r'[^\w\s]', '', lower_text)

print("\n2. LOWERCASE TEXT:")
print(lower_text)

print("\nCLEANED TEXT (WITHOUT PUNCTUATION):")
print(clean_text)


# STEP 4: Tokenize into Sentences and Words
sentences = sent_tokenize(text)
words = word_tokenize(clean_text)

print("\n3. SENTENCE TOKENIZATION:")
for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sentence}")

print("Total Sentences:", len(sentences))

print("\nWORD TOKENIZATION:")
print(words)
print("Total Word Tokens:", len(words))


# STEP 5: Compare split() and word_tokenize()
split_tokens = lower_text.split()
nltk_tokens = word_tokenize(lower_text)

print("\n4. COMPARISON: split() vs word_tokenize()")

print("\nPython split() Output:")
print(split_tokens)
print("Number of tokens:", len(split_tokens))

print("\nNLTK word_tokenize() Output:")
print(nltk_tokens)
print("Number of tokens:", len(nltk_tokens))

print("\nDifference:")
print("split() separates text based on whitespace.")
print("word_tokenize() also identifies punctuation separately.")


# STEP 6: Remove Stopwords
stop_words = set(stopwords.words('english'))

# Identify stopwords
removed_stopwords = [
    word for word in words
    if word in stop_words
]

# Remove stopwords
filtered_words = [
    word for word in words
    if word not in stop_words
]

print("\n5. STOPWORD REMOVAL")

print("Stopwords Found:")
print(removed_stopwords)

print("\nWords After Stopword Removal:")
print(filtered_words)

print("\nTotal Words Before Removal:", len(words))
print("Total Words After Removal:", len(filtered_words))


# STEP 7: Word Frequency Distribution
frequency = FreqDist(filtered_words)

print("\n6. WORD FREQUENCY DISTRIBUTION:")
for word, count in frequency.most_common():
    print(f"{word}: {count}")

# Display top 10 frequent words
print("\nTOP 10 MOST FREQUENT WORDS:")
print(frequency.most_common(10))

# Plot frequency distribution
plt.figure(figsize=(10, 5))
frequency.plot(10, title="Top 10 Most Frequent Words")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.show()


# STEP 8: Final Summary
print("\n" + "=" * 60)
print("FINAL SUMMARY")
print("=" * 60)

print("Total Sentences:", len(sentences))
print("Total Word Tokens:", len(words))
print("Stopwords Removed:", len(removed_stopwords))
print("Final Words:", len(filtered_words))
print("Unique Final Words:", len(set(filtered_words)))


```

Q2. Using the same paragraph from Q1:
1. Extract all words with only alphabets using re.findall()
2. Remove stop words using NLTK’s stopword list
3. Perform stemming with PorterStemmer
4. Perform lemma5za5on with WordNetLemma5zer
5. Compare the stemmed and lemma5zed outputs and explain when you’d prefer one over
the other.

```python

import re
import nltk

from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag

# Download required resources
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)

# STEP 1: Use the same paragraph from Q1
text = """
Artificial intelligence is changing the way people work and learn.
It helps computers understand information and solve complex problems.
AI is used in healthcare, education, transportation, and many other fields.
Machine learning allows systems to improve their performance using data.
AI tools can save time and support people in making better decisions.
I believe artificial intelligence will play an important role in the future.
"""

print("=" * 60)
print("Q2: STEMMING AND LEMMATIZATION")
print("=" * 60)

# STEP 2: Convert to lowercase and extract alphabet-only words
text = text.lower()
tokens = re.findall(r'\b[a-zA-Z]+\b', text)

print("\n1. Alphabet-only words:")
print(tokens)
print("Total words:", len(tokens))

# STEP 3: Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in tokens
    if word not in stop_words
]

print("\n2. Words after stopword removal:")
print(filtered_words)
print("Total words after removal:", len(filtered_words))

# STEP 4: Perform Stemming using PorterStemmer
porter = PorterStemmer()

stemmed_words = [
    porter.stem(word)
    for word in filtered_words
]

print("\n3. Porter Stemmer Output:")
print(stemmed_words)

# STEP 5: Perform POS-aware Lemmatization
lemmatizer = WordNetLemmatizer()

# Convert NLTK POS tags into WordNet POS tags
def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

tagged_words = pos_tag(filtered_words)

lemmatized_words = [
    lemmatizer.lemmatize(word, get_wordnet_pos(tag))
    for word, tag in tagged_words
]

print("\n4. WordNet Lemmatizer Output:")
print(lemmatized_words)

# STEP 6: Compare Stemming and Lemmatization
print("\n5. COMPARISON TABLE")
print("-" * 50)
print(f"{'Original Word':<20}{'Stemmed':<15}{'Lemmatized'}")
print("-" * 50)

for original, stem, lemma in zip(
    filtered_words, stemmed_words, lemmatized_words
):
    print(f"{original:<20}{stem:<15}{lemma}")

# STEP 7: Explanation
print("\n6. OBSERVATION:")
print("Stemming removes word endings using simple rules.")
print("It may produce words that are not dictionary words.")
print("Lemmatization converts words into meaningful dictionary forms.")

print("\nWhen to prefer Stemming:")
print("When speed and reducing word variations are important.")

print("\nWhen to prefer Lemmatization:")
print("When meaningful and grammatically correct words are important.")


```

Q3. Choose 3 short texts of your own (e.g., different news headlines, product reviews).
1. Use CountVectorizer to generate the Bag of Words representa5on.
2. Use TfidfVectorizer to compute TF-IDF scores.
3. Print and interpret the top 3 keywords from each text using TF-IDF.

```python


import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

print("=" * 65)
print("Q3: BAG OF WORDS AND TF-IDF")
print("=" * 65)

# STEP 1: Create three short texts
texts = [
    "Artificial intelligence helps computers learn and solve problems. AI improves healthcare, education and technology.",
    "Blockchain technology provides secure digital transactions. Blockchain stores data in distributed and transparent networks.",
    "Machine learning enables computers to learn from data. Machine learning helps build intelligent prediction models."
]

text_names = ["Text 1 - Artificial Intelligence",
              "Text 2 - Blockchain",
              "Text 3 - Machine Learning"]

print("\n1. ORIGINAL TEXTS:")
for name, text in zip(text_names, texts):
    print(f"\n{name}:")
    print(text)


# STEP 2: Bag of Words using CountVectorizer
count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

feature_names = count_vectorizer.get_feature_names_out()

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=feature_names,
    index=text_names
)

print("\n2. BAG OF WORDS REPRESENTATION:")
print(bow_df.to_string())


# STEP 3: TF-IDF using TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

tfidf_features = tfidf_vectorizer.get_feature_names_out()

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_features,
    index=text_names
)

print("\n3. TF-IDF REPRESENTATION:")
print(tfidf_df.round(3).to_string())


# STEP 4: Find Top 3 Keywords from Each Text
print("\n4. TOP 3 KEYWORDS USING TF-IDF:")
print("-" * 50)

for i, name in enumerate(text_names):

    scores = tfidf_matrix[i].toarray().flatten()

    top_indices = scores.argsort()[-3:][::-1]

    print(f"\n{name}:")

    for index in top_indices:
        print(
            f"{tfidf_features[index]} : "
            f"{scores[index]:.4f}"
        )


# STEP 5: Interpretation
print("\n5. INTERPRETATION:")

print("""
Text 1:
The important keywords represent Artificial Intelligence
and its applications in healthcare and education.

Text 2:
The keywords represent Blockchain, digital transactions,
distributed storage and network security.

Text 3:
The keywords represent Machine Learning, data and
prediction models.

Conclusion:
CountVectorizer counts word occurrences, whereas
TF-IDF identifies words based on their importance
in a particular document compared to other documents.
""")


```

Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs
Blockchain).
1. Preprocess and tokenize both texts.
2. Calculate:
a. Jaccard Similarity using sets
b. Cosine Similarity using TfidfVectorizer + cosine_similarity()
c. Analyze which similarity metric gives be^er insights in your case.

```python


import re
import nltk

from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download('stopwords', quiet=True)

print("=" * 65)
print("Q4: TEXT SIMILARITY")
print("=" * 65)


# STEP 1: Write two texts about different technologies

text1 = """
Artificial intelligence enables computers to perform intelligent tasks.
AI systems learn from data and identify useful patterns.
Artificial intelligence is widely used in healthcare and education.
Machine learning helps computers make predictions and decisions.
AI improves automation and supports problem solving.
"""

text2 = """
Blockchain is a technology used for secure digital transactions.
It stores information in distributed and connected blocks.
Blockchain provides transparency and data security.
Cryptocurrency uses blockchain for digital payments.
Blockchain removes the need for a central authority.
"""

print("\n1. ORIGINAL TEXTS")

print("\nText 1: Artificial Intelligence")
print(text1)

print("\nText 2: Blockchain Technology")
print(text2)


# STEP 2: Preprocess and tokenize both texts

stop_words = set(stopwords.words('english'))

def preprocess(text):
    # Convert text into lowercase
    text = text.lower()

    # Extract alphabet-only words
    tokens = re.findall(r'\b[a-z]+\b', text)

    # Remove stopwords
    tokens = [word for word in tokens if word not in stop_words]

    return tokens


tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("\n2. PREPROCESSING AND TOKENIZATION")

print("\nAI Tokens:")
print(tokens1)

print("\nBlockchain Tokens:")
print(tokens2)


# STEP 3: Calculate Jaccard Similarity

set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard_score = len(intersection) / len(union) if union else 0

print("\n3. JACCARD SIMILARITY")

print("Common Words:", intersection)
print("Number of Common Words:", len(intersection))
print("Total Unique Words:", len(union))

print("Jaccard Similarity:", round(jaccard_score, 4))


# STEP 4: Calculate Cosine Similarity using TF-IDF

processed_text1 = " ".join(tokens1)
processed_text2 = " ".join(tokens2)

documents = [processed_text1, processed_text2]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

cosine_score = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]

print("\n4. COSINE SIMILARITY")

print("TF-IDF Matrix:")
print(tfidf_matrix.toarray().round(3))

print("Cosine Similarity:", round(cosine_score, 4))


# STEP 5: Compare Both Similarity Metrics

print("\n5. COMPARISON")

print("Jaccard Similarity:", round(jaccard_score, 4))
print("Cosine Similarity:", round(cosine_score, 4))

if jaccard_score > cosine_score:
    print("\nJaccard Similarity gives a higher score.")
elif cosine_score > jaccard_score:
    print("\nCosine Similarity gives a higher score.")
else:
    print("\nBoth metrics give the same score.")


# STEP 6: Final Analysis

print("\n6. ANALYSIS")

print("""
Jaccard Similarity compares the common unique words
between two texts. It does not consider word frequency.

Cosine Similarity compares TF-IDF vectors and considers
the importance of words in each text.

In this example, both technologies are different, but
they may share some general technical vocabulary.

Jaccard is useful for comparing word overlap, whereas
Cosine Similarity is useful for comparing document
content using TF-IDF representations.
""")
```

Q5. Write a short review for a product or service.
1. Use TextBlob or VADER to find polarity & subjec5vity for each review.
2. Classify reviews into Posi5ve / Nega5ve / Neutral.
3. Create a word cloud using the wordcloud library for all posi5ve reviews.

```python


# STEP 1: Install and import required libraries
!pip -q install textblob wordcloud

import pandas as pd
import matplotlib.pyplot as plt

from textblob import TextBlob
from wordcloud import WordCloud, STOPWORDS

print("=" * 65)
print("Q5: SENTIMENT ANALYSIS AND WORD CLOUD")
print("=" * 65)


# STEP 2: Create sample smartphone reviews
reviews = [
    "This smartphone has an excellent camera and amazing battery life.",
    "I love this phone because it is fast and easy to use.",
    "The display quality is beautiful and the performance is great.",
    "The battery drains quickly and the phone becomes very slow.",
    "I am disappointed with the poor camera and bad sound quality.",
    "The phone is expensive and the customer service is terrible.",
    "The phone comes with a charger and a USB cable.",
    "It has a large display and comes in three different colours."
]

print("\n1. ORIGINAL REVIEWS")

for i, review in enumerate(reviews, start=1):
    print(f"Review {i}: {review}")


# STEP 3: Calculate Polarity and Subjectivity
results = []

for review in reviews:

    analysis = TextBlob(review)

    polarity = analysis.sentiment.polarity
    subjectivity = analysis.sentiment.subjectivity

    # Classify sentiment
    if polarity > 0.1:
        sentiment = "Positive"
    elif polarity < -0.1:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    results.append({
        "Review": review,
        "Polarity": round(polarity, 3),
        "Subjectivity": round(subjectivity, 3),
        "Sentiment": sentiment
    })


# STEP 4: Display sentiment analysis results
df = pd.DataFrame(results)

print("\n2. SENTIMENT ANALYSIS RESULTS")
print(df.to_string(index=False))


# STEP 5: Count Positive, Negative and Neutral Reviews
sentiment_counts = df["Sentiment"].value_counts()

print("\n3. SENTIMENT CLASSIFICATION SUMMARY")
print(sentiment_counts.to_string())


# STEP 6: Collect all positive reviews
positive_reviews = df[
    df["Sentiment"] == "Positive"
]["Review"].tolist()

print("\n4. POSITIVE REVIEWS")

for review in positive_reviews:
    print("-", review)


# STEP 7: Create Word Cloud from positive reviews
positive_text = " ".join(positive_reviews)

if positive_text.strip():

    wordcloud = WordCloud(
        width=1000,
        height=500,
        background_color="white",
        stopwords=STOPWORDS,
        collocations=False
    ).generate(positive_text)

    plt.figure(figsize=(12, 6))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Word Cloud of Positive Smartphone Reviews")
    plt.show()

else:
    print("No positive reviews available to create Word Cloud.")


# STEP 8: Final Summary
print("\n5. FINAL SUMMARY")
print("Total Reviews:", len(reviews))
print("Positive Reviews:", (df["Sentiment"] == "Positive").sum())
print("Negative Reviews:", (df["Sentiment"] == "Negative").sum())
print("Neutral Reviews:", (df["Sentiment"] == "Neutral").sum())


```

Q6. Choose your own paragraph (~100 words) as training data.
1. Tokenize text using Tokenizer() from keras.preprocessing.text
2. Create input sequences and build a simple LSTM or Dense model
3. Train the model and generate 2–3 new lines of text star5ng from any seed word you
provide.

```python


import numpy as np
import tensorflow as tf
import random

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Input
from tensorflow.keras.utils import to_categorical

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)
random.seed(42)

print("=" * 65)
print("Q6: TEXT GENERATION USING LSTM")
print("=" * 65)


# STEP 1: Create approximately 100-word training paragraph

text = """
Artificial intelligence is transforming the modern world.
It allows computers to learn from data and perform intelligent tasks.
Machine learning is an important part of artificial intelligence.
It helps systems identify patterns and make useful predictions.
Technology is improving healthcare, education and transportation.
Smart computers can analyse information and solve complex problems.
Deep learning uses neural networks to understand complicated data.
Artificial intelligence saves time and improves productivity.
Students use technology to learn new skills and explore ideas.
In the future, intelligent machines may help people in many ways.
Artificial intelligence continues to create new opportunities for society.
Learning about technology helps us prepare for the future.
"""

print("\n1. TRAINING DATA:")
print(text)


# STEP 2: Tokenize text using Keras Tokenizer

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("\n2. TOKENIZATION:")
print("Total Unique Words:", total_words - 1)
print("Word Index:")
print(tokenizer.word_index)


# STEP 3: Create input sequences

input_sequences = []

for line in text.split('.'):
    line = line.strip()

    if not line:
        continue

    token_list = tokenizer.texts_to_sequences([line])[0]

    # Create sequences of increasing length
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i + 1]
        input_sequences.append(n_gram_sequence)

# Find maximum sequence length
max_sequence_len = max(
    len(sequence) for sequence in input_sequences
)

# Apply padding
input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding='pre'
)

print("\n3. INPUT SEQUENCES:")
print("Total Input Sequences:", len(input_sequences))
print("Maximum Sequence Length:", max_sequence_len)
print("Sample Padded Sequences:")
print(input_sequences[:5])


# STEP 4: Prepare input X and output y

X = input_sequences[:, :-1]
y = input_sequences[:, -1]

# Convert output into categorical format
y = to_categorical(y, num_classes=total_words)

print("\n4. TRAINING DATA PREPARATION:")
print("X Shape:", X.shape)
print("y Shape:", y.shape)


# STEP 5: Build LSTM Model

model = Sequential([
    Input(shape=(max_sequence_len - 1,)),
    Embedding(total_words, 64),
    LSTM(100),
    Dense(total_words, activation='softmax')
])

model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

print("\n5. LSTM MODEL SUMMARY:")
model.summary()


# STEP 6: Train the Model

print("\n6. MODEL TRAINING:")

history = model.fit(
    X,
    y,
    epochs=150,
    verbose=0
)

print("Model training completed successfully!")

print("Final Training Accuracy:",
      round(history.history['accuracy'][-1] * 100, 2), "%")


# STEP 7: Generate New Text using a Seed Word

def generate_text(seed_text, next_words=8):

    for _ in range(next_words):

        token_list = tokenizer.texts_to_sequences(
            [seed_text]
        )[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding='pre'
        )

        predicted_probabilities = model.predict(
            token_list,
            verbose=0
        )[0]

        predicted_word_index = np.argmax(
            predicted_probabilities
        )

        output_word = None

        for word, index in tokenizer.word_index.items():
            if index == predicted_word_index:
                output_word = word
                break

        if output_word is None:
            break

        seed_text += " " + output_word

    return seed_text


# STEP 8: Generate 3 New Lines

print("\n7. GENERATED TEXT:")

seed_words = ["artificial", "technology", "machine"]

for i, seed in enumerate(seed_words, start=1):

    generated_line = generate_text(
        seed_text=seed,
        next_words=8
    )

    print(f"Generated Line {i}: {generated_line}")


# STEP 9: Final Summary

print("\n" + "=" * 65)
print("FINAL SUMMARY")
print("=" * 65)

print("Total Unique Words:", total_words - 1)
print("Input Sequences:", len(input_sequences))
print("Maximum Sequence Length:", max_sequence_len)
print("Model: LSTM")
print("Epochs: 150")
print("Generated Lines: 3")

```
