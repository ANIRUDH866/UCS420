## Q1. Sensor Readings

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected = temperature + 2
print("Corrected temperatures:", corrected)

fahrenheit = (9/5) * corrected + 32
print("Fahrenheit:", fahrenheit)

above_32 = temperature[temperature > 32]
print("Readings greater than 32°C:", above_32)

count = np.sum(temperature > 32)
print("Count:", count)
```

Vectorization applies an operation to the complete array at once, while Boolean indexing selects values that satisfy a condition. This avoids writing a separate loop for each reading and makes NumPy operations efficient.

## Q2. Daily Steps

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", steps.sum())
print("Mean steps:", steps.mean())
print("Maximum:", steps.max())
print("Minimum:", steps.min())
print("Total for each day:", steps.sum(axis=0))
print("Total for each user:", steps.sum(axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum position:", max_position)
```

## Q3. NumPy Slicing, Copying and Reshaping

```python
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999

print("Original:", original)
print("Subset:", subset)

copied = original[1:5].copy()
copied[0] = 500

print("Original after copy modification:", original)
print("Copied array:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("3 x 4 matrix:")
print(matrix)

print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[1:3, 2:4])

flat = matrix.flatten()
ravelled = matrix.ravel()

print("Flatten:", flat)
print("Ravel:", ravelled)

ravelled[0] = 100
print("After changing ravel:")
print("Matrix:", matrix)
print("Ravel:", ravelled)

flat[1] = 200
print("After changing flatten:")
print("Matrix:", matrix)
print("Flatten:", flat)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

The slice `subset` shares data with `original`, so changing it changes the original array. The `.copy()` slice has its own data, so changing it does not change the original array. `ravel()` normally returns a view when possible, while `flatten()` returns a copy.

## Q4. Cognitive-Assistive System

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T
print("X.T:")
print(XT)

XTX = XT @ X
print("X.T @ X:")
print(XTX)

XTX_inv = np.linalg.inv(XTX)
print("Inverse of X.T @ X:")
print(XTX_inv)

beta = XTX_inv @ XT @ y
print("Coefficients:")
print(beta)

new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("Predicted assistance score:", prediction)
```

The three coefficients correspond to sleep hours, activity level, and stress level. Each coefficient indicates the change in the predicted assistance score associated with that feature when the other features are held constant.
